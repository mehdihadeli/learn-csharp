## LINQ to XML

LINQ to XML turns XML into an in-memory object graph that you can query and modify with ordinary C# code.

See the full lesson in [../../docs/09-linq-and-data-queries/06-linq-to-xml.md](../../docs/09-linq-and-data-queries/06-linq-to-xml.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [LINQ to XML overview](https://learn.microsoft.com/en-us/dotnet/standard/linq/linq-xml-overview).

### Example

```python
using System.Xml.Linq;

XDocument document = new(
    new XElement("books",
        new XElement("book", new XAttribute("id", 1), new XElement("title", "C# Basics")),
        new XElement("book", new XAttribute("id", 2), new XElement("title", "LINQ in Practice"))));

var titles = document.Root!
    .Elements("book")
    .Select(book => (string?)book.Element("title"));

Console.WriteLine(string.Join(", ", titles));
```

LINQ to XML is especially readable when the document is small enough to load into memory and you want query-like navigation.
