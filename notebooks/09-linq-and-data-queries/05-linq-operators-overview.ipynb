## LINQ Operators Overview

Most LINQ queries combine a few operator families: filtering, projection, ordering, grouping, and materialization.

See the full lesson in [../../docs/09-linq-and-data-queries/05-linq-operators-overview.md](../../docs/09-linq-and-data-queries/05-linq-operators-overview.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Standard query operators overview](https://learn.microsoft.com/en-us/dotnet/csharp/linq/standard-query-operators/).

### Example

```python
var products = new[]
{
    new { Name = "Book", Category = "Reading", Price = 30m },
    new { Name = "Notebook", Category = "Stationery", Price = 8m },
    new { Name = "Pen", Category = "Stationery", Price = 4m }
};

var query = products
    .Where(product => product.Price >= 5m)
    .OrderBy(product => product.Category)
    .ThenBy(product => product.Name)
    .Select(product => $"{product.Category}: {product.Name} = {product.Price:C}")
    .ToList();

Console.WriteLine(string.Join(Environment.NewLine, query));
```

This one pipeline uses `Where`, `OrderBy`, `ThenBy`, `Select`, and `ToList`, which is a good pattern for recognizing operator families in practice.
