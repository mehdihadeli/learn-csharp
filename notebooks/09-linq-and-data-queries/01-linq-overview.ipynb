## LINQ Overview

LINQ treats data processing as a pipeline of filtering, projection, ordering, and aggregation steps instead of a series of manual loops.

See the full lesson in [../../docs/09-linq-and-data-queries/01-linq-overview.md](../../docs/09-linq-and-data-queries/01-linq-overview.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Introduction to LINQ Queries in C#](https://learn.microsoft.com/en-us/dotnet/csharp/linq/get-started/introduction-to-linq-queries).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
var products = new[]
{
    new { Name = "Keyboard", Price = 120m },
    new { Name = "Mouse", Price = 35m },
    new { Name = "Monitor", Price = 240m }
};

var premiumNames = products
    .Where(product => product.Price >= 100m)
    .Select(product => product.Name);

Console.WriteLine(string.Join(", ", premiumNames));
```

Production note: LINQ is strongest when each step communicates intent clearly and avoids hiding expensive work.
