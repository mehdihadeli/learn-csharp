## LINQ OverviewLINQ treats data processing as a pipeline of filtering, projection, ordering, and aggregation steps instead of a series of manual loops.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
var products = new[]{    new { Name = "Keyboard", Price = 120m },    new { Name = "Mouse", Price = 35m },    new { Name = "Monitor", Price = 240m }};var premiumNames = products    .Where(product => product.Price >= 100m)    .Select(product => product.Name);Console.WriteLine(string.Join(", ", premiumNames));
```
