## IQueryable, Expression Trees, and ProvidersAn `IQueryable<T>` carries an expression tree that a provider can inspect and translate, unlike a plain in-memory `IEnumerable<T>` pipeline.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
using System.Linq.Expressions;var values = new[] { 1, 2, 3, 4, 5 };IQueryable<int> query = values.AsQueryable().Where(number => number > 2);Console.WriteLine(query.Expression);Console.WriteLine(string.Join(", ", query));
```

Production note: understanding the materialization boundary matters because the code before it may be translated by a provider instead of running locally line by line.### PracticeAdd a `Select` clause and inspect how the expression tree changes.
