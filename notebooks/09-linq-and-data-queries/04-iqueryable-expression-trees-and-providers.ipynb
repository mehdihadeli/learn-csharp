## IQueryable, Expression Trees, and Providers

An `IQueryable<T>` carries an expression tree that a provider can inspect and translate, unlike a plain in-memory `IEnumerable<T>` pipeline.

See the full lesson in [../../docs/09-linq-and-data-queries/04-iqueryable-expression-trees-and-providers.md](../../docs/09-linq-and-data-queries/04-iqueryable-expression-trees-and-providers.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [IQueryable](https://learn.microsoft.com/en-us/dotnet/api/system.linq.iqueryable-1) and [Expression trees](https://learn.microsoft.com/en-us/dotnet/csharp/advanced-topics/expression-trees/).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using System.Linq.Expressions;

var values = new[] { 1, 2, 3, 4, 5 };
IQueryable<int> query = values.AsQueryable().Where(number => number > 2);

Console.WriteLine(query.Expression);
Console.WriteLine(string.Join(", ", query));
```

Production note: understanding the materialization boundary matters because the code before it may be translated by a provider instead of running locally line by line.
