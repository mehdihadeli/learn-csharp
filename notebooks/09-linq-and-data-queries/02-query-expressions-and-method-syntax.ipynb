## Query Expressions and Method SyntaxC# supports both SQL-like query expressions and method-chain syntax, and the compiler translates query expressions into method calls.See the full lesson in [../../docs/09-linq-and-data-queries/02-query-expressions-and-method-syntax.md](../../docs/09-linq-and-data-queries/02-query-expressions-and-method-syntax.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn LINQ for C#](https://learn.microsoft.com/dotnet/csharp/linq/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
var numbers = new[] { 1, 2, 3, 4, 5, 6 };var queryExpression =    from number in numbers    where number % 2 == 0    select number * 10;var methodSyntax = numbers    .Where(number => number % 2 == 0)    .Select(number => number * 10);Console.WriteLine($"Query syntax:  {string.Join(", ", queryExpression)}");Console.WriteLine($"Method syntax: {string.Join(", ", methodSyntax)}");
```
