## Query Expressions and Method Syntax

C# supports both SQL-like query expressions and method-chain syntax, and the compiler translates query expressions into method calls.

See the full lesson in [../../docs/09-linq-and-data-queries/02-query-expressions-and-method-syntax.md](../../docs/09-linq-and-data-queries/02-query-expressions-and-method-syntax.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Query expression basics](https://learn.microsoft.com/en-us/dotnet/csharp/linq/get-started/query-expression-basics).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
var numbers = new[] { 1, 2, 3, 4, 5, 6 };

var queryExpression =
    from number in numbers
    where number % 2 == 0
    select number * 10;

var methodSyntax = numbers
    .Where(number => number % 2 == 0)
    .Select(number => number * 10);

Console.WriteLine($"Query syntax:  {string.Join(", ", queryExpression)}");
Console.WriteLine($"Method syntax: {string.Join(", ", methodSyntax)}");
```

Production note: choose the style that makes the transformation easiest to read, and mix styles only when that improves clarity.
