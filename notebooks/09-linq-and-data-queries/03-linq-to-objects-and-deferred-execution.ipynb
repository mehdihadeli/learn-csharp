## LINQ to Objects and Deferred ExecutionDeferred execution means many LINQ queries are not evaluated when you define them, but when you enumerate them.See the full lesson in [../../docs/09-linq-and-data-queries/03-linq-to-objects-and-deferred-execution.md](../../docs/09-linq-and-data-queries/03-linq-to-objects-and-deferred-execution.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn LINQ for C#](https://learn.microsoft.com/dotnet/csharp/linq/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
var numbers = new List<int> { 1, 2, 3 };var query = numbers.Where(number => number >= 2);numbers.Add(4);Console.WriteLine(string.Join(", ", query));Console.WriteLine("Materialized snapshot: " + string.Join(", ", query.ToList()));
```
