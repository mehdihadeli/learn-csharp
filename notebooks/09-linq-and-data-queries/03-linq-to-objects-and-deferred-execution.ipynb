## LINQ to Objects and Deferred Execution

Deferred execution means many LINQ queries are not evaluated when you define them, but when you enumerate them.

See the full lesson in [../../docs/09-linq-and-data-queries/03-linq-to-objects-and-deferred-execution.md](../../docs/09-linq-and-data-queries/03-linq-to-objects-and-deferred-execution.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Introduction to LINQ Queries in C#](https://learn.microsoft.com/en-us/dotnet/csharp/linq/get-started/introduction-to-linq-queries#classification-of-standard-query-operators-by-manner-of-execution).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
var numbers = new List<int> { 1, 2, 3 };
var query = numbers.Where(number => number >= 2);

numbers.Add(4);

Console.WriteLine(string.Join(", ", query));
Console.WriteLine("Materialized snapshot: " + string.Join(", ", query.ToList()));
```

Production note: deferred execution is useful, but repeated enumeration can surprise you when the source changes or the query is expensive.
