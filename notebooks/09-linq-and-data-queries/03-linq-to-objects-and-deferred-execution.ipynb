## LINQ to Objects and Deferred ExecutionDeferred execution means many LINQ queries are not evaluated when you define them, but when you enumerate them.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
var numbers = new List<int> { 1, 2, 3 };var query = numbers.Where(number => number >= 2);numbers.Add(4);Console.WriteLine(string.Join(", ", query));Console.WriteLine("Materialized snapshot: " + string.Join(", ", query.ToList()));
```

Production note: deferred execution is useful, but repeated enumeration can surprise you when the source changes or the query is expensive.### PracticeCreate one deferred query and one materialized list from the same source, then mutate the source and compare the results.
