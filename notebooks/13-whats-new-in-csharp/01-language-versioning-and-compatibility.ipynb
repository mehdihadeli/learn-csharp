## Language Versioning and Compatibility

Language versioning is really a compatibility policy question involving the SDK, compiler, target framework, and team build environment together.

See the full lesson in [../../docs/13-whats-new-in-csharp/01-language-versioning-and-compatibility.md](../../docs/13-whats-new-in-csharp/01-language-versioning-and-compatibility.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [C# language versioning](https://learn.microsoft.com/en-us/dotnet/csharp/language-reference/language-versioning).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using System.Xml.Linq;

var project = XElement.Parse(@"<Project><PropertyGroup><TargetFramework>net10.0</TargetFramework><LangVersion>preview</LangVersion></PropertyGroup></Project>");

Console.WriteLine($"Target framework: {project.Descendants("TargetFramework").Single().Value}");
Console.WriteLine($"Language version: {project.Descendants("LangVersion").Single().Value}");
Console.WriteLine("Use explicit versions deliberately when reproducibility matters.");
```

Production note: `latest` and `preview` should be policy decisions, not casual defaults, because they affect reproducibility and team consistency.
