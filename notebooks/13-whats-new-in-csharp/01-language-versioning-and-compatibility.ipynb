## Language Versioning and CompatibilityLanguage versioning is really a compatibility policy question involving the SDK, compiler, target framework, and team build environment together.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
using System.Xml.Linq;var project = XElement.Parse(@"<Project><PropertyGroup><TargetFramework>net10.0</TargetFramework><LangVersion>preview</LangVersion></PropertyGroup></Project>");Console.WriteLine($"Target framework: {project.Descendants("TargetFramework").Single().Value}");Console.WriteLine($"Language version: {project.Descendants("LangVersion").Single().Value}");Console.WriteLine("Use explicit versions deliberately when reproducibility matters.");
```
