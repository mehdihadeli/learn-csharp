### Example

```polyglot-notebook
var features = new[]
{
    new { Feature = "params collections", EverydayUse = true },
    new { Feature = "System.Threading.Lock support", EverydayUse = true },
    new { Feature = "allows ref struct constraints", EverydayUse = false },
    new { Feature = "ref struct interface support", EverydayUse = false }
};

foreach (var feature in features.Where(item => item.EverydayUse))
{
    Console.WriteLine(feature.Feature);
}
```
