## Strings and Text Processing

Text handling usually means normalizing, splitting, comparing, and formatting strings in a deliberate way rather than applying random methods one by one.

See the full lesson in [../../docs/11-csharp-how-to-guides/01-strings-and-text-processing.md](../../docs/11-csharp-how-to-guides/01-strings-and-text-processing.md) for the broader explanation and surrounding context.
Original Microsoft Learn references: [Strings in .NET](https://learn.microsoft.com/en-us/dotnet/standard/base-types/strings) and [StringBuilder](https://learn.microsoft.com/en-us/dotnet/standard/base-types/stringbuilder).
### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
var input = "  Ada, grace , LINA  ";

var normalizedNames = input
    .Split(',')
    .Select(name => name.Trim())
    .Where(name => name.Length > 0)
    .Select(name => name.ToUpperInvariant())
    .ToArray();

Console.WriteLine(string.Join(" | ", normalizedNames));
```

Production note: choose explicit comparison and normalization rules so technical identifiers and display text are not handled accidentally the same way.
