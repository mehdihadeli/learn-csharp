## Delegates, Events, and Command-Line Input

Small console tools often combine three ideas at once: input from outside the program, delegates for behavior, and events for notifications.

See the full lesson in [../../docs/11-csharp-how-to-guides/05-delegates-events-and-command-line-input.md](../../docs/11-csharp-how-to-guides/05-delegates-events-and-command-line-input.md) for the broader explanation and surrounding context.

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
Action<string> write = Console.WriteLine;
write("Hello from a delegate");

var processor = new Processor();
processor.Completed += () => Console.WriteLine("Processing finished.");

var args = new[] { "report.txt" };
processor.Run(args[0], write);

class Processor
{
    public event Action? Completed;

    public void Run(string input, Action<string> writeOutput)
    {
        writeOutput($"Processing: {input}");
        Completed?.Invoke();
    }
}
```

Production note: command-line arguments are external input, so validate them even in small internal tools.
