## Collections and Initializers

Initializers make setup declarative and readable when object creation is simple and the chosen values should be visible together.

See the full lesson in [../../docs/11-csharp-how-to-guides/04-collections-and-initializers.md](../../docs/11-csharp-how-to-guides/04-collections-and-initializers.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Object and collection initializers](https://learn.microsoft.com/en-us/dotnet/csharp/programming-guide/classes-and-structs/object-and-collection-initializers).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
var course = new Course
{
    Title = "Learn C#",
    DurationInHours = 12,
    Tags = { "Language", "Platform" }
};

Console.WriteLine(course.Title);
Console.WriteLine(string.Join(", ", course.Tags));

class Course
{
    public string Title { get; set; } = string.Empty;
    public int DurationInHours { get; set; }
    public List<string> Tags { get; } = new();
}
```

Production note: prefer constructors or factories when important invariants must be enforced at creation time.
