## Collections and InitializersInitializers make setup declarative and readable when object creation is simple and the chosen values should be visible together.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
var course = new Course{    Title = "Learn C#",    DurationInHours = 12,    Tags = { "Language", "Platform" }};Console.WriteLine(course.Title);Console.WriteLine(string.Join(", ", course.Tags));class Course{    public string Title { get; set; } = string.Empty;    public int DurationInHours { get; set; }    public List<string> Tags { get; } = new();}
```

Production note: prefer constructors or factories when important invariants must be enforced at creation time.### PracticeAdd another object initializer and decide whether a constructor would communicate the intent better.
