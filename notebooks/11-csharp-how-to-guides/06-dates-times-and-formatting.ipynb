## Dates, Times, and Formatting

Date and time code becomes much easier to reason about when you separate timestamps, local display values, and durations.

See the full lesson in [../../docs/11-csharp-how-to-guides/06-dates-times-and-formatting.md](../../docs/11-csharp-how-to-guides/06-dates-times-and-formatting.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Date and time in .NET](https://learn.microsoft.com/en-us/dotnet/standard/datetime/) and [Standard date and time format strings](https://learn.microsoft.com/en-us/dotnet/standard/base-types/standard-date-and-time-format-strings).

### Example

```python
using System.Globalization;

DateTimeOffset timestamp = DateTimeOffset.UtcNow;
string serialized = timestamp.ToString("O", CultureInfo.InvariantCulture);

Console.WriteLine($"Round-trip: {serialized}");

if (DateTimeOffset.TryParse(serialized, CultureInfo.InvariantCulture, DateTimeStyles.RoundtripKind, out var parsed))
{
    Console.WriteLine($"Display date: {parsed:yyyy-MM-dd}");
}
```

Use invariant round-trip formats for storage and transport, and friendlier formats only when you are presenting values to people.
