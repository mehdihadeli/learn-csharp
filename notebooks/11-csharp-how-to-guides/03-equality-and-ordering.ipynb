## Equality and OrderingReference identity, value equality, and ordering answer different questions, and good domain modeling depends on keeping them distinct.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
public sealed record Person(string Name, int Age);var left = new Person("Lina", 30);var right = new Person("Lina", 30);var ages = new[] { 30, 18, 24 }.OrderBy(age => age);Console.WriteLine(ReferenceEquals(left, right));Console.WriteLine(left == right);Console.WriteLine(string.Join(", ", ages));
```

Production note: equality rules affect dictionaries, sets, tests, and business logic, so they should follow domain meaning rather than convenience alone.### PracticeCreate a `Money` record and decide which fields should matter for equality.
