## Equality and OrderingReference identity, value equality, and ordering answer different questions, and good domain modeling depends on keeping them distinct.See the full lesson in [../../docs/11-csharp-how-to-guides/03-equality-and-ordering.md](../../docs/11-csharp-how-to-guides/03-equality-and-ordering.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn C# how-to guides](https://learn.microsoft.com/dotnet/csharp/how-to/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
public sealed record Person(string Name, int Age);var left = new Person("Lina", 30);var right = new Person("Lina", 30);var ages = new[] { 30, 18, 24 }.OrderBy(age => age);Console.WriteLine(ReferenceEquals(left, right));Console.WriteLine(left == right);Console.WriteLine(string.Join(", ", ages));
```
