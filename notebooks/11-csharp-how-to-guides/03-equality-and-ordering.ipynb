## Equality and Ordering

Reference identity, value equality, and ordering answer different questions, and good domain modeling depends on keeping them distinct.

See the full lesson in [../../docs/11-csharp-how-to-guides/03-equality-and-ordering.md](../../docs/11-csharp-how-to-guides/03-equality-and-ordering.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Compare strings in .NET](https://learn.microsoft.com/en-us/dotnet/standard/base-types/comparing) and [Best practices for comparing strings in .NET](https://learn.microsoft.com/en-us/dotnet/standard/base-types/best-practices-strings).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
public sealed record Person(string Name, int Age);

var left = new Person("Lina", 30);
var right = new Person("Lina", 30);
var ages = new[] { 30, 18, 24 }.OrderBy(age => age);

Console.WriteLine(ReferenceEquals(left, right));
Console.WriteLine(left == right);
Console.WriteLine(string.Join(", ", ages));
```

Production note: equality rules affect dictionaries, sets, tests, and business logic, so they should follow domain meaning rather than convenience alone.
