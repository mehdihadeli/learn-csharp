## Analyzers and Code FixesAnalyzers turn coding guidance into diagnostics, and code fixes turn those diagnostics into guided repairs.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
#r "nuget: Microsoft.CodeAnalysis.CSharp, 4.13.0"using System.Linq;using Microsoft.CodeAnalysis;using Microsoft.CodeAnalysis.CSharp;using Microsoft.CodeAnalysis.CSharp.Syntax;var descriptor = new DiagnosticDescriptor(    id: "DEMO001",    title: "Avoid var in this demo",    messageFormat: "Use an explicit type instead of 'var' in this demo",    category: "Style",    defaultSeverity: DiagnosticSeverity.Info,    isEnabledByDefault: true);var tree = CSharpSyntaxTree.ParseText("class Demo { void Run() { var value = 42; } }");var varNode = tree.GetRoot()    .DescendantNodes()    .OfType<IdentifierNameSyntax>()    .First(node => node.Identifier.Text == "var");var diagnostic = Diagnostic.Create(descriptor, varNode.GetLocation());Console.WriteLine($"{diagnostic.Id}: {diagnostic.GetMessage()}");
```
