## Analyzers and Code Fixes

Analyzers turn coding guidance into diagnostics, and code fixes turn those diagnostics into guided repairs.

See the full lesson in [../../docs/12-roslyn-and-code-analysis/03-analyzers-and-code-fixes.md](../../docs/12-roslyn-and-code-analysis/03-analyzers-and-code-fixes.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Tutorial: Write your first analyzer and code fix](https://learn.microsoft.com/en-us/dotnet/csharp/roslyn-sdk/tutorials/how-to-write-csharp-analyzer-code-fix).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
#r "nuget: Microsoft.CodeAnalysis.CSharp, 4.13.0"
using System.Linq;
using Microsoft.CodeAnalysis;
using Microsoft.CodeAnalysis.CSharp;
using Microsoft.CodeAnalysis.CSharp.Syntax;

var descriptor = new DiagnosticDescriptor(
    id: "DEMO001",
    title: "Avoid var in this demo",
    messageFormat: "Use an explicit type instead of 'var' in this demo",
    category: "Style",
    defaultSeverity: DiagnosticSeverity.Info,
    isEnabledByDefault: true);

var tree = CSharpSyntaxTree.ParseText("class Demo { void Run() { var value = 42; } }");
var varNode = tree.GetRoot()
    .DescendantNodes()
    .OfType<IdentifierNameSyntax>()
    .First(node => node.Identifier.Text == "var");

var diagnostic = Diagnostic.Create(descriptor, varNode.GetLocation());
Console.WriteLine($"{diagnostic.Id}: {diagnostic.GetMessage()}");
```

Production note: a useful analyzer rule needs both accurate detection and a message that helps developers act on it quickly.
