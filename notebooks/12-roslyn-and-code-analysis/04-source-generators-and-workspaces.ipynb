## Source Generators and Workspaces

Source generators contribute code during compilation, while workspaces model projects and documents for broader tooling tasks.

See the full lesson in [../../docs/12-roslyn-and-code-analysis/04-source-generators-and-workspaces.md](../../docs/12-roslyn-and-code-analysis/04-source-generators-and-workspaces.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [The .NET Compiler Platform SDK: Source generators](https://learn.microsoft.com/en-us/dotnet/csharp/roslyn-sdk/#source-generators) and [Work with a workspace](https://learn.microsoft.com/en-us/dotnet/csharp/roslyn-sdk/work-with-workspace).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
#r "nuget: Microsoft.CodeAnalysis.CSharp.Workspaces, 4.13.0"
using System.Linq;
using Microsoft.CodeAnalysis;
using Microsoft.CodeAnalysis.Text;

var workspace = new AdhocWorkspace();
var project = workspace.AddProject("NotebookDemo", LanguageNames.CSharp);
var document = workspace.AddDocument(project.Id, "Demo.cs", SourceText.From("class Demo { }"));

Console.WriteLine(workspace.CurrentSolution.Projects.Count());
Console.WriteLine(document.Name);
```

Production note: choose workspaces when the task is understanding or changing code across files, not generating new code during compilation.
