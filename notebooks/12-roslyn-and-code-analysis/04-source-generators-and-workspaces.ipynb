## Source Generators and WorkspacesSource generators contribute code during compilation, while workspaces model projects and documents for broader tooling tasks.See the full lesson in [../../docs/12-roslyn-and-code-analysis/04-source-generators-and-workspaces.md](../../docs/12-roslyn-and-code-analysis/04-source-generators-and-workspaces.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn Roslyn SDK](https://learn.microsoft.com/dotnet/csharp/roslyn-sdk/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
#r "nuget: Microsoft.CodeAnalysis.CSharp.Workspaces, 4.13.0"using System.Linq;using Microsoft.CodeAnalysis;using Microsoft.CodeAnalysis.Text;var workspace = new AdhocWorkspace();var project = workspace.AddProject("NotebookDemo", LanguageNames.CSharp);var document = workspace.AddDocument(project.Id, "Demo.cs", SourceText.From("class Demo { }"));Console.WriteLine(workspace.CurrentSolution.Projects.Count());Console.WriteLine(document.Name);
```

Production note: choose workspaces when the task is understanding or changing code across files, not generating new code during compilation.Return to [the full lesson](../../docs/12-roslyn-and-code-analysis/04-source-generators-and-workspaces.md) when you want the fuller explanation, tradeoffs, and surrounding examples.Official Microsoft Learn reference: [Microsoft Learn Roslyn SDK](https://learn.microsoft.com/dotnet/csharp/roslyn-sdk/).### PracticeAdd a second document to the workspace and inspect the project document count.
