## Syntax and Semantic AnalysisSyntax tells you the shape of code, while semantics tells you what symbols and types that code actually refers to.See the full lesson in [../../docs/12-roslyn-and-code-analysis/02-syntax-and-semantic-analysis.md](../../docs/12-roslyn-and-code-analysis/02-syntax-and-semantic-analysis.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn Roslyn SDK](https://learn.microsoft.com/dotnet/csharp/roslyn-sdk/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
#r "nuget: Microsoft.CodeAnalysis.CSharp, 4.13.0"using System.Linq;using Microsoft.CodeAnalysis;using Microsoft.CodeAnalysis.CSharp;using Microsoft.CodeAnalysis.CSharp.Syntax;var tree = CSharpSyntaxTree.ParseText("using System; class Demo { void Run() { string name = Console.ReadLine(); } }");var compilation = CSharpCompilation.Create("Demo")    .AddSyntaxTrees(tree)    .AddReferences(        MetadataReference.CreateFromFile(typeof(object).Assembly.Location),        MetadataReference.CreateFromFile(typeof(Console).Assembly.Location),        MetadataReference.CreateFromFile(typeof(Enumerable).Assembly.Location));var semanticModel = compilation.GetSemanticModel(tree);var declaration = tree.GetRoot().DescendantNodes().OfType<VariableDeclarationSyntax>().Single();Console.WriteLine($"Syntax type text: {declaration.Type}");Console.WriteLine($"Semantic type: {semanticModel.GetTypeInfo(declaration.Type).Type?.ToDisplayString()}");
```
