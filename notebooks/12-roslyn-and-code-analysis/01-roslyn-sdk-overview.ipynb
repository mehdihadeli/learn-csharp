## Roslyn SDK Overview

Roslyn exposes the compiler model so tools can inspect syntax, resolve semantics, and build refactorings, analyzers, or generators.

See the full lesson in [../../docs/12-roslyn-and-code-analysis/01-roslyn-sdk-overview.md](../../docs/12-roslyn-and-code-analysis/01-roslyn-sdk-overview.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [.NET Compiler Platform SDK](https://learn.microsoft.com/en-us/dotnet/csharp/roslyn-sdk/).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
#r "nuget: Microsoft.CodeAnalysis.CSharp, 4.13.0"
using Microsoft.CodeAnalysis.CSharp;

var tree = CSharpSyntaxTree.ParseText("int total = 1 + 2;");
Console.WriteLine(tree.GetRoot().Kind());
```
