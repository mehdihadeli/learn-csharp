## Caller Information Attributes

Caller information attributes let the compiler fill in call-site details such as member name and line number.

See the full lesson in [../../docs/07-advanced-language-features/12-caller-information-attributes.md](../../docs/07-advanced-language-features/12-caller-information-attributes.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Caller information](https://learn.microsoft.com/en-us/dotnet/csharp/language-reference/attributes/caller-information).

### Example

```python
using System.Runtime.CompilerServices;

static void Log(string message,
    [CallerMemberName] string memberName = "",
    [CallerLineNumber] int lineNumber = 0)
{
    Console.WriteLine($"[{memberName}:{lineNumber}] {message}");
}

void ProcessOrder()
{
    Log("Starting order processing");
}

ProcessOrder();
```

Caller info reduces fragile hard-coded strings in helpers, logs, and validation utilities.
