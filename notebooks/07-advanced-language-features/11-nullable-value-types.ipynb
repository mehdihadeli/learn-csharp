## Nullable Value Types

Nullable value types model the difference between a real value and the absence of a value.

See the full lesson in [../../docs/07-advanced-language-features/11-nullable-value-types.md](../../docs/07-advanced-language-features/11-nullable-value-types.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Nullable value types](https://learn.microsoft.com/en-us/dotnet/csharp/language-reference/builtin-types/nullable-value-types).

### Example

```python
int? discountPercent = null;
int appliedDiscount = discountPercent ?? 10;

Console.WriteLine($"Discount has value: {discountPercent.HasValue}");
Console.WriteLine($"Applied discount: {appliedDiscount}%");

discountPercent = 25;
Console.WriteLine($"Updated discount: {discountPercent}%");
```

Use `T?` when absence is a real part of the domain, not as a vague fallback for uncertain design.
