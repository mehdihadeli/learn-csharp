## Async Exceptions and Task CompositionAsync failures appear when you await the task, so exception handling belongs around the awaited operation rather than around task creation alone.See the full lesson in [../../docs/10-asynchronous-programming/04-async-exceptions-and-task-composition.md](../../docs/10-asynchronous-programming/04-async-exceptions-and-task-composition.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn asynchronous programming in C#](https://learn.microsoft.com/dotnet/csharp/asynchronous-programming/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
async Task<int> GoodAsync(){    await Task.Delay(25);    return 5;}async Task<int> BadAsync(){    await Task.Delay(25);    throw new InvalidOperationException("Simulated failure");}try{    await Task.WhenAll(GoodAsync(), BadAsync());}catch (Exception exception){    Console.WriteLine(exception.Message);}
```
