## Composing Async WorkflowsAsync workflows stay readable when you decompose them into small helpers and keep each awaited step responsible for one clear action.See the full lesson in [../../docs/10-asynchronous-programming/03-composing-async-workflows.md](../../docs/10-asynchronous-programming/03-composing-async-workflows.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn asynchronous programming in C#](https://learn.microsoft.com/dotnet/csharp/asynchronous-programming/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
async Task<string> ReadNameAsync(string raw){    await Task.Delay(50);    return raw.Trim();}async Task<string> NormalizeAsync(string name){    await Task.Delay(50);    return name.ToUpperInvariant();}var normalized = await NormalizeAsync(await ReadNameAsync("  ada lovelace  "));Console.WriteLine(normalized);
```

Production note: small, named async helpers are easier to test and reason about than one large method with several unrelated awaits.Return to [the full lesson](../../docs/10-asynchronous-programming/03-composing-async-workflows.md) when you want the fuller explanation, tradeoffs, and surrounding examples.Official Microsoft Learn reference: [Microsoft Learn asynchronous programming in C#](https://learn.microsoft.com/dotnet/csharp/asynchronous-programming/).### PracticeAdd a third helper that formats the final value for display and keep the workflow readable.
