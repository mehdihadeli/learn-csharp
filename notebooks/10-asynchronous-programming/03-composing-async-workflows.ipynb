## Composing Async Workflows

Async workflows stay readable when you decompose them into small helpers and keep each awaited step responsible for one clear action.

See the full lesson in [../../docs/10-asynchronous-programming/03-composing-async-workflows.md](../../docs/10-asynchronous-programming/03-composing-async-workflows.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Task asynchronous programming model](https://learn.microsoft.com/en-us/dotnet/csharp/asynchronous-programming/task-asynchronous-programming-model).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
async Task<string> ReadNameAsync(string raw)
{
    await Task.Delay(50);
    return raw.Trim();
}

async Task<string> NormalizeAsync(string name)
{
    await Task.Delay(50);
    return name.ToUpperInvariant();
}

var normalized = await NormalizeAsync(await ReadNameAsync("  ada lovelace  "));
Console.WriteLine(normalized);
```

Production note: small, named async helpers are easier to test and reason about than one large method with several unrelated awaits.
