## Starting and Coordinating Tasks

Task coordination becomes useful when several independent operations can run before you await their combined result.

See the full lesson in [../../docs/10-asynchronous-programming/02-starting-and-coordinating-tasks.md](../../docs/10-asynchronous-programming/02-starting-and-coordinating-tasks.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Task asynchronous programming model](https://learn.microsoft.com/en-us/dotnet/csharp/asynchronous-programming/task-asynchronous-programming-model).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
async Task<int> LoadValueAsync(int delay, int value)
{
    await Task.Delay(delay);
    return value;
}

var results = await Task.WhenAll(
    LoadValueAsync(150, 10),
    LoadValueAsync(100, 20),
    LoadValueAsync(50, 30));

Console.WriteLine(string.Join(", ", results));
Console.WriteLine($"Total = {results.Sum()}");
```

Production note: start independent work before awaiting it when concurrency is safe and the operations do not depend on each other.
