## Starting and Coordinating TasksTask coordination becomes useful when several independent operations can run before you await their combined result.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
async Task<int> LoadValueAsync(int delay, int value){    await Task.Delay(delay);    return value;}var results = await Task.WhenAll(    LoadValueAsync(150, 10),    LoadValueAsync(100, 20),    LoadValueAsync(50, 30));Console.WriteLine(string.Join(", ", results));Console.WriteLine($"Total = {results.Sum()}");
```

Production note: start independent work before awaiting it when concurrency is safe and the operations do not depend on each other.### PracticeAdd a `Task.WhenAny` example and print which operation completes first.
