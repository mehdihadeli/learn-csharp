## Starting and Coordinating TasksTask coordination becomes useful when several independent operations can run before you await their combined result.See the full lesson in [../../docs/10-asynchronous-programming/02-starting-and-coordinating-tasks.md](../../docs/10-asynchronous-programming/02-starting-and-coordinating-tasks.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn asynchronous programming in C#](https://learn.microsoft.com/dotnet/csharp/asynchronous-programming/).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
async Task<int> LoadValueAsync(int delay, int value){    await Task.Delay(delay);    return value;}var results = await Task.WhenAll(    LoadValueAsync(150, 10),    LoadValueAsync(100, 20),    LoadValueAsync(50, 30));Console.WriteLine(string.Join(", ", results));Console.WriteLine($"Total = {results.Sum()}");
```
