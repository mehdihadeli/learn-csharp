## async and await FundamentalsAsync code lets a program represent ongoing work without blocking the current flow while it waits for I/O or other operations to complete.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
using System.Diagnostics;var stopwatch = Stopwatch.StartNew();await Task.Delay(150);stopwatch.Stop();Console.WriteLine($"Elapsed: {stopwatch.ElapsedMilliseconds} ms");
```

Production note: async is most valuable for wait-heavy work such as file, database, and network operations, not for every line of code automatically.### PracticeWrap the delay in a helper method that returns `Task<string>` and await it from the notebook cell.
