## Security, Cryptography, and Safe Code

Security is a cross-cutting concern that includes inputs, secrets, dependencies, cryptography choices, and safer platform defaults.

### Example

```polyglot-notebook
using System.Security.Cryptography;

var bytes = RandomNumberGenerator.GetBytes(16);
var token = Convert.ToBase64String(bytes);

Console.WriteLine(token);
Console.WriteLine($"Length: {bytes.Length} bytes");
```

Production note: secure code is usually the result of disciplined boundaries and good defaults, not one isolated security feature.
