## Logging, Diagnostics, and Observability

Observability helps you answer runtime questions such as what happened, how long it took, and where the failure came from.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/08-logging-diagnostics-and-observability.md](../../docs/14-dotnet-platform-and-runtime/08-logging-diagnostics-and-observability.md) for the broader explanation and surrounding context.

### Real-world angle

Use this notebook to inspect a small diagnostic workflow that combines structured logging, `Stopwatch`, and stack-trace capture.

### Example

```polyglot-notebook
#r "nuget: Microsoft.Extensions.Logging.Console, 9.0.0"
using Microsoft.Extensions.Logging;
using System.Diagnostics;

using var loggerFactory = LoggerFactory.Create(builder => builder.AddConsole());
var logger = loggerFactory.CreateLogger("NotebookDemo");
var stopwatch = Stopwatch.StartNew();

try
{
    logger.LogInformation("Starting diagnostic sample at {Time}", DateTimeOffset.UtcNow);
    throw new InvalidOperationException("Synthetic failure for notebook diagnostics");
}
catch (Exception exception)
{
    logger.LogError(exception, "Captured {ExceptionType}", exception.GetType().Name);
    var stackTrace = new StackTrace(exception, true);
    Console.WriteLine(stackTrace);
}
finally
{
    stopwatch.Stop();
    logger.LogInformation("Elapsed milliseconds: {Elapsed}", stopwatch.ElapsedMilliseconds);
}
```

Production note: logs tell you what the application recorded, a stack trace tells you where the failure happened, and timing data helps explain performance behavior.
