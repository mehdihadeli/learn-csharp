## Runtime, CLR, and Managed Execution

Managed execution means the runtime participates actively in loading code, handling exceptions, tracking objects, and executing compiled instructions.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/02-runtime-clr-and-managed-execution.md](../../docs/14-dotnet-platform-and-runtime/02-runtime-clr-and-managed-execution.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Common Language Runtime overview](https://learn.microsoft.com/en-us/dotnet/standard/clr) and [Managed execution process](https://learn.microsoft.com/en-us/dotnet/standard/managed-execution-process).
