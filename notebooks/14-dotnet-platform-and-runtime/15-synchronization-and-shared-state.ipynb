## Synchronization and Shared State

Once multiple execution paths can touch the same mutable state, correctness depends on coordination as much as on code structure.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/15-synchronization-and-shared-state.md](../../docs/14-dotnet-platform-and-runtime/15-synchronization-and-shared-state.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Microsoft Learn Introduction to .NET](https://learn.microsoft.com/dotnet/core/introduction).

### Real-world angle

Use this notebook to see a tiny synchronized component where access to shared state stays behind one gate.

### Example

```python
using System.Threading.Tasks;

var gate = new object();
var counter = 0;

Parallel.For(0, 1000, _ =>
{
    lock (gate)
    {
        counter++;
    }
});

Console.WriteLine($"Counter = {counter}");
```

Production note: `lock` is useful, but the cleaner design often comes from reducing shared mutable state instead of protecting more and more of it.
