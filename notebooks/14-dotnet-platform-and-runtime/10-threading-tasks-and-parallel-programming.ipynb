## Threading, Tasks, and Parallel ProgrammingConcurrency tools solve different categories of problems, so choosing between async, parallelism, and synchronization begins with understanding the workload.### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
using var source = new CancellationTokenSource(TimeSpan.FromSeconds(1));var results = await Task.WhenAll(    Task.Run(() => 10, source.Token),    Task.Run(() => 20, source.Token));Console.WriteLine(string.Join(", ", results));Console.WriteLine($"Total = {results.Sum()}");
```
