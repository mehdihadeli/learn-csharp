## Threading, Tasks, and Parallel Programming

Concurrency tools solve different categories of problems, so choosing between async, parallelism, and synchronization begins with understanding the workload.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/10-threading-tasks-and-parallel-programming.md](../../docs/14-dotnet-platform-and-runtime/10-threading-tasks-and-parallel-programming.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Microsoft Learn Introduction to .NET](https://learn.microsoft.com/dotnet/core/introduction).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using var source = new CancellationTokenSource(TimeSpan.FromSeconds(1));

var results = await Task.WhenAll(
    Task.Run(() => 10, source.Token),
    Task.Run(() => 20, source.Token));

Console.WriteLine(string.Join(", ", results));
Console.WriteLine($"Total = {results.Sum()}");
```

Production note: tasks are the main application-level abstraction for work, but they are not the same thing as raw threads.
