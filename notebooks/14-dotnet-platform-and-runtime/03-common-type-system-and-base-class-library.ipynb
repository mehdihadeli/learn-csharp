## Common Type System and Base Class Library

The .NET type system gives runtime meaning to types, while the base class library supplies the APIs most C# programs rely on every day.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/03-common-type-system-and-base-class-library.md](../../docs/14-dotnet-platform-and-runtime/03-common-type-system-and-base-class-library.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Common type system](https://learn.microsoft.com/en-us/dotnet/standard/base-types/common-type-system) and [Introduction to .NET: Components](https://learn.microsoft.com/en-us/dotnet/core/introduction#components).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
int count = 3;
System.Int32 sameCount = count;
var names = new List<string> { "Ada", "Grace" };

Console.WriteLine(count == sameCount);
Console.WriteLine(names.Count);
```
