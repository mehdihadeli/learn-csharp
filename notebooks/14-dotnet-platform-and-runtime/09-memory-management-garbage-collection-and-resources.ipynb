## Memory Management, Garbage Collection, and Resources

Managed memory cleanup and explicit external resource cleanup are related but not identical responsibilities.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/09-memory-management-garbage-collection-and-resources.md](../../docs/14-dotnet-platform-and-runtime/09-memory-management-garbage-collection-and-resources.md) for the broader explanation and surrounding context.

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using var stream = new MemoryStream();
using var writer = new StreamWriter(stream);

writer.Write("managed memory and resource cleanup");
writer.Flush();

Console.WriteLine(stream.Length);
```

Production note: garbage collection handles unreachable managed objects, but `IDisposable` exists because files, sockets, and handles need prompt cleanup.
