## .NET Platform Overview

.NET is the runtime, library, and tooling platform around C#, so understanding the platform means understanding more than language syntax.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/01-dotnet-platform-overview.md](../../docs/14-dotnet-platform-and-runtime/01-dotnet-platform-overview.md) for the broader explanation and surrounding context.
Original Microsoft Learn reference: [Microsoft Learn Introduction to .NET](https://learn.microsoft.com/dotnet/core/introduction).
### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using System.Runtime.InteropServices;

Console.WriteLine($"Framework: {RuntimeInformation.FrameworkDescription}");
Console.WriteLine($"OS: {RuntimeInformation.OSDescription}");
Console.WriteLine($"Architecture: {RuntimeInformation.ProcessArchitecture}");
```

Production note: platform details such as runtime version and operating system often explain behavior that source code alone does not.
