## Assemblies, Reflection, and Metadata

Assemblies package code and metadata, and reflection is how .NET inspects that metadata at runtime.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/18-assemblies-reflection-and-metadata.md](../../docs/14-dotnet-platform-and-runtime/18-assemblies-reflection-and-metadata.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Assemblies in .NET](https://learn.microsoft.com/en-us/dotnet/standard/assembly/) and [Reflection and attributes](https://learn.microsoft.com/en-us/dotnet/csharp/advanced-topics/reflection-and-attributes/).

### Example

```python
using System.Reflection;
using System.Text;

Assembly assembly = typeof(StringBuilder).Assembly;
Console.WriteLine($"Assembly: {assembly.GetName().Name}");

Type type = typeof(StringBuilder);
var propertyNames = type.GetProperties().Select(property => property.Name).Take(5);
Console.WriteLine(string.Join(", ", propertyNames));
```

Reflection is great for discovery and tooling, but it should solve a real runtime-discovery problem rather than replace direct type usage everywhere.
