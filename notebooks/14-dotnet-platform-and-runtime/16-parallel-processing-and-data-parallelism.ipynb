## Parallel Processing and Data Parallelism

Parallelism is mainly about CPU-bound work across multiple cores, which is a different performance problem from async I/O.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/16-parallel-processing-and-data-parallelism.md](../../docs/14-dotnet-platform-and-runtime/16-parallel-processing-and-data-parallelism.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Data parallelism](https://learn.microsoft.com/en-us/dotnet/standard/parallel-programming/data-parallelism-task-parallel-library) and [Parallel LINQ (PLINQ)](https://learn.microsoft.com/en-us/dotnet/standard/parallel-programming/introduction-to-plinq).

### Real-world angle

Use this notebook to try a small PLINQ transformation and connect it to the idea of independent CPU work.

### Example

```python
using System.Linq;

var results = Enumerable.Range(1, 12)
    .AsParallel()
    .Select(n => new { Number = n, Square = n * n })
    .OrderBy(x => x.Number)
    .ToList();

foreach (var item in results)
{
    Console.WriteLine($"{item.Number} -> {item.Square}");
}
```

Production note: `AsParallel()` can be expressive, but it only helps when the workload is truly CPU-bound, sufficiently large, and mostly independent.
