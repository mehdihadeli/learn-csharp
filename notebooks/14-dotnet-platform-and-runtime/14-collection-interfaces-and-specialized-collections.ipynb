## Collection Interfaces and Specialized Collections

Collection interfaces communicate what callers can do, while specialized collections such as queues, concurrent collections, and immutable collections communicate behavioral intent.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/14-collection-interfaces-and-specialized-collections.md](../../docs/14-dotnet-platform-and-runtime/14-collection-interfaces-and-specialized-collections.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Collections and data structures](https://learn.microsoft.com/en-us/dotnet/standard/collections/) and [Commonly used collection types](https://learn.microsoft.com/en-us/dotnet/standard/collections/selecting-a-collection-class).

### Real-world angle

Use this notebook to compare API contracts and processing order with a small runnable example.

### Example

```python
using System.Collections.Concurrent;

var internalList = new List<string> { "parse", "validate", "publish" };
IReadOnlyList<string> steps = internalList;

Console.WriteLine($"First step from IReadOnlyList: {steps[0]}");

var queue = new Queue<string>();
queue.Enqueue("job-1");
queue.Enqueue("job-2");
Console.WriteLine($"Queue gives oldest item first: {queue.Dequeue()}");

var concurrentQueue = new ConcurrentQueue<string>();
concurrentQueue.Enqueue("work-a");
concurrentQueue.Enqueue("work-b");

if (concurrentQueue.TryDequeue(out var item))
{
    Console.WriteLine($"ConcurrentQueue dequeued: {item}");
}
```

Production note: a good API often stores data in one concrete collection internally but exposes a narrower interface such as `IEnumerable<T>` or `IReadOnlyList<T>` externally.
