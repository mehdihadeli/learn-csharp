## Networking, URIs, and HttpClient

URIs identify resources, and `HttpClient` is the main high-level API for HTTP communication in .NET.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/17-networking-uris-and-httpclient.md](../../docs/14-dotnet-platform-and-runtime/17-networking-uris-and-httpclient.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [System.Uri supplementary remarks](https://learn.microsoft.com/en-us/dotnet/fundamentals/runtime-libraries/system-uri) and [Guidelines for using HttpClient](https://learn.microsoft.com/en-us/dotnet/fundamentals/networking/http/httpclient-guidelines).

### Example

```python
Uri docsUri = new("https://learn.microsoft.com/en-us/dotnet/");
HttpRequestMessage request = new(HttpMethod.Get, docsUri);
request.Headers.Add("User-Agent", "learn-csharp-notebook");

Console.WriteLine($"Scheme: {docsUri.Scheme}");
Console.WriteLine($"Host: {docsUri.Host}");
Console.WriteLine($"Path: {docsUri.AbsolutePath}");
Console.WriteLine($"Method: {request.Method}");
```

This notebook keeps the example offline and deterministic, but it still shows the relationship between a URI and an HTTP request message.
