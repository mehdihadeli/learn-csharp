## Configuration, Dependency Injection, and Generic Host

Configuration, dependency injection, and hosting are usually coordinated together in modern .NET applications instead of being isolated topics.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/07-configuration-dependency-injection-and-generic-host.md](../../docs/14-dotnet-platform-and-runtime/07-configuration-dependency-injection-and-generic-host.md) for the broader explanation and surrounding context.

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
#r "nuget: Microsoft.Extensions.DependencyInjection, 9.0.0"
using Microsoft.Extensions.DependencyInjection;

var services = new ServiceCollection();
services.AddSingleton<IClock, SystemClock>();

using var provider = services.BuildServiceProvider();
var clock = provider.GetRequiredService<IClock>();
Console.WriteLine(clock.GetLabel());

interface IClock
{
    string GetLabel();
}

sealed class SystemClock : IClock
{
    public string GetLabel() => "Resolved through DI";
}
```

Production note: use DI when abstractions, lifetimes, and configuration-driven behavior matter, not just to add ceremony around every dependency.
