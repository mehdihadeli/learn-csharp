## Streams and I/O

Streams provide the shared .NET abstraction for byte-oriented I/O, while readers and writers add text-oriented behavior on top.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/13-streams-and-io.md](../../docs/14-dotnet-platform-and-runtime/13-streams-and-io.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [File and stream I/O](https://learn.microsoft.com/en-us/dotnet/standard/io/).

### Real-world angle

Use this notebook to observe the difference between raw bytes and text-oriented wrappers without depending on a real file path.

### Example

```python
using System.Text;

await using var stream = new MemoryStream();
await using (var writer = new StreamWriter(stream, Encoding.UTF8, leaveOpen: true))
{
    await writer.WriteLineAsync("Hello from a stream");
    await writer.WriteLineAsync("Text writers sit on top of bytes");
    await writer.FlushAsync();
}

Console.WriteLine($"Stream length in bytes: {stream.Length}");

stream.Position = 0;
using var reader = new StreamReader(stream, Encoding.UTF8, leaveOpen: true);
var text = await reader.ReadToEndAsync();

Console.WriteLine(text);
```

Production note: many high-level APIs eventually rely on streams underneath, so understanding streams helps connect file, network, compression, and serialization APIs.
