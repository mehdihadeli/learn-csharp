## .NET SDK, Project System, and MSBuildThe SDK and project system provide the conventions that make modern `.csproj` files short while still driving restore, build, test, and publish behavior.See the full lesson in [../../docs/14-dotnet-platform-and-runtime/05-dotnet-sdk-project-system-and-msbuild.md](../../docs/14-dotnet-platform-and-runtime/05-dotnet-sdk-project-system-and-msbuild.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn Introduction to .NET](https://learn.microsoft.com/dotnet/core/introduction).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
using System.Xml.Linq;var project = XElement.Parse(@"<Project Sdk=""Microsoft.NET.Sdk""><PropertyGroup><OutputType>Exe</OutputType><TargetFramework>net10.0</TargetFramework><Nullable>enable</Nullable></PropertyGroup></Project>");foreach (var element in project.Descendants().Where(element => !element.HasElements)){    Console.WriteLine($"{element.Name}: {element.Value}");}
```
