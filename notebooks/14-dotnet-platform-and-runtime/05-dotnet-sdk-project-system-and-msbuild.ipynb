## .NET SDK, Project System, and MSBuild

The SDK and project system provide the conventions that make modern `.csproj` files short while still driving restore, build, test, and publish behavior.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/05-dotnet-sdk-project-system-and-msbuild.md](../../docs/14-dotnet-platform-and-runtime/05-dotnet-sdk-project-system-and-msbuild.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [.NET project SDKs](https://learn.microsoft.com/en-us/dotnet/core/project-sdk/overview) and [MSBuild](https://learn.microsoft.com/en-us/visualstudio/msbuild/msbuild?view=vs-2022).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using System.Xml.Linq;

var project = XElement.Parse(@"<Project Sdk=""Microsoft.NET.Sdk""><PropertyGroup><OutputType>Exe</OutputType><TargetFramework>net10.0</TargetFramework><Nullable>enable</Nullable></PropertyGroup></Project>");

foreach (var element in project.Descendants().Where(element => !element.HasElements))
{
    Console.WriteLine($"{element.Name}: {element.Value}");
}
```

Production note: a short project file usually means the SDK is supplying conventions for you, not that the build pipeline is trivial.
