## NuGet Packages and Dependency Management

Package management is really dependency management, which means compatibility, security, and reproducibility all matter when you add a package.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/06-nuget-packages-and-dependency-management.md](../../docs/14-dotnet-platform-and-runtime/06-nuget-packages-and-dependency-management.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [What is NuGet?](https://learn.microsoft.com/en-us/nuget/what-is-nuget).

### Example

```polyglot-notebook
using System.Xml.Linq;

var packageReference = XElement.Parse(@"<PackageReference Include=""Humanizer"" Version=""2.14.1"" />");
Console.WriteLine(packageReference.Attribute("Include")?.Value);
Console.WriteLine(packageReference.Attribute("Version")?.Value);
```

Production note: package updates are engineering decisions, not only convenience actions, because dependencies affect build and runtime behavior.
