## Target Frameworks, TFMs, and .NET Standard

A target framework is a compatibility statement about APIs, runtimes, and consumers, not just a short string in a project file.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/04-target-frameworks-tfms-and-dotnet-standard.md](../../docs/14-dotnet-platform-and-runtime/04-target-frameworks-tfms-and-dotnet-standard.md) for the broader explanation and surrounding context.

Original Microsoft Learn references: [Target frameworks in SDK-style projects](https://learn.microsoft.com/en-us/dotnet/standard/frameworks) and [.NET Standard](https://learn.microsoft.com/en-us/dotnet/standard/net-standard).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
using System.Xml.Linq;

var project = XElement.Parse(@"<Project><PropertyGroup><TargetFrameworks>net8.0;net10.0</TargetFrameworks></PropertyGroup></Project>");
var frameworks = project.Descendants("TargetFrameworks").Single().Value.Split(';');

Console.WriteLine(string.Join(" | ", frameworks));
```

Production note: multi-targeting is usually a library decision driven by consumer compatibility rather than something every application needs.
