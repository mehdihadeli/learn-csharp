## Target Frameworks, TFMs, and .NET StandardA target framework is a compatibility statement about APIs, runtimes, and consumers, not just a short string in a project file.See the full lesson in [../../docs/14-dotnet-platform-and-runtime/04-target-frameworks-tfms-and-dotnet-standard.md](../../docs/14-dotnet-platform-and-runtime/04-target-frameworks-tfms-and-dotnet-standard.md) for the broader explanation and surrounding context.Original Microsoft Learn reference: [Microsoft Learn Introduction to .NET](https://learn.microsoft.com/dotnet/core/introduction).### Real-world angleUse this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.### Example

```
using System.Xml.Linq;var project = XElement.Parse(@"<Project><PropertyGroup><TargetFrameworks>net8.0;net10.0</TargetFrameworks></PropertyGroup></Project>");var frameworks = project.Descendants("TargetFrameworks").Single().Value.Split(';');Console.WriteLine(string.Join(" | ", frameworks));
```
