## Regular Expressions

Regular expressions are a compact way to describe text shape for matching, extraction, replacement, and splitting.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/19-regular-expressions.md](../../docs/14-dotnet-platform-and-runtime/19-regular-expressions.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [Regular expressions in .NET](https://learn.microsoft.com/en-us/dotnet/standard/base-types/regular-expressions).
