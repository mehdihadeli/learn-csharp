## Publishing, Deployment, and Versioning

Publishing is a deployment decision that changes portability, size, runtime requirements, and compatibility expectations.

See the full lesson in [../../docs/14-dotnet-platform-and-runtime/11-publishing-deployment-and-versioning.md](../../docs/14-dotnet-platform-and-runtime/11-publishing-deployment-and-versioning.md) for the broader explanation and surrounding context.

Original Microsoft Learn reference: [.NET application publishing overview](https://learn.microsoft.com/en-us/dotnet/core/deploying/).

### Real-world angle

Use this notebook to run a focused example, observe the output, and connect the concept to ordinary .NET development work.

### Example

```polyglot-notebook
var publishModes = new[]
{
    new { Mode = "framework-dependent", Tradeoff = "smaller output, runtime required on target" },
    new { Mode = "self-contained", Tradeoff = "larger output, runtime bundled" },
    new { Mode = "single-file", Tradeoff = "simpler distribution, different packaging tradeoffs" }
};

foreach (var mode in publishModes)
{
    Console.WriteLine($"{mode.Mode}: {mode.Tradeoff}");
}
```

Production note: a successful build does not guarantee a successful deployment, because the target environment and publish settings still matter.
